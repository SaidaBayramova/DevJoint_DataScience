# Checkpoint 1 : Framing the Problem



## 1. Business problem

The dataset contains real estate listings scraped from bina.az

Sellers and agents on the platform currently set listing prices manually, which means prices are inconsistent. Similar properties end up listed at very different prices just because of guesswork.

We want an automated way to estimate a fair market price for a property, based on its features, instead of relying on manual guesswork.

it can helps sellers/agents price new listings realistically instead of guessing, flags listings that are clearly overpriced or underpriced compared to similar properties, useful for buyers browsing the platform and for the platform's own moderation, could support a "suggested price range" feature shown directly on the listing page.


## 2. Target variable

The target is price. The listing price of the property. This is a regression problem, not classification, since price can be any positive value, not a fixed set of categories.

All listings are already priced in a single currency (AZN), so no currency conversion is needed here, unlike some other datasets. The minimum is 11 AZN and the maximum is 600,000,000 AZN, neither of which is a realistic sale price.

## 3. Success metrics

Since this is regression, accuracy doesn't apply — the right metrics measure how far off a price prediction is:

MAE (Mean Absolute Error), in AZN. The main metric to report to the business partner, since "the model is off by about X AZN on average" is easy for a non-technical person to understand.

RMSE (Root Mean Squared Error), in AZN. Reported alongside MAE, since RMSE punishes large mistakes more. A big gap between MAE and RMSE would mean the model is occasionally very wrong on some listings, which matters for a pricing tool.

R2, a secondary, at-a-glance number for comparing models against each other during development, not something to lead with in the business report.

MAPE (Mean Absolute Percentage Error), reported as a percentage, because "the estimate is typically within X% of the real price" is often more intuitive for a business audience than a raw AZN figure, especially across a wide price range.
