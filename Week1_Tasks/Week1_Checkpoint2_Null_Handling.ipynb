**Goal:** Define and apply a separate, reasoned strategy for each of the null values. Instead of handling all gaps the same way, I investigated why each column is missing, random (MCAR) or systematic (MNAR), and based my decision on that.

## 1. Reloading the dataset

```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', 60)
pd.set_option('display.width', 150)

df = pd.read_csv('house_sale.csv')
print(df.shape)
```

## 2. Overview of missing values

```python
null_counts = df.isnull().sum()
null_percent = (null_counts / len(df) * 100).round(2)
null_summary = pd.DataFrame({"null_count": null_counts, "null_percent": null_percent}).sort_values("null_percent", ascending = False)
null_summary[null_summary["null_count"] > 0]
```

## 3. Investigation: MCAR or MNAR?

Before dropping or filling anything, I checked whether each column's missingness pattern is related to another column.

```python
print("Missing rate of 'Mərtəbə' (floor) by Kateqoriya")
df.groupby("Kateqoriya")["Mərtəbə"].apply(lambda x: x.isnull().mean().round(3))
```

```python
print("Missing rate of 'Torpaq sahəsi' (land area) by Kateqoriya")
df.groupby("Kateqoriya")["Torpaq sahəsi"].apply(lambda x: x.isnull().mean().round(3))
```

```python
print("Missing rate of 'Otaq sayı' (number of rooms) by Kateqoriya")
df.groupby("Kateqoriya")["Otaq sayı"].apply(lambda x: x.isnull().mean().round(3))
```

**Result:**
- "Mərtəbə" (floor) is only populated for "Köhnə tikili" and "Yeni tikili"; it's 100% missing for every other category (House/Garden, Object, Office, Garage, Land), which makes logical sense, since "floor number" only applies to apartments. This is a textbook MNAR (structural missing) pattern.
- "Torpaq sahəsi" (land area) is the opposite, only populated for "Həyət evi"/"Bağ evi" (house/garden, 0% missing), 100% missing everywhere else. Again MNAR, land area only applies to houses.
- "Otaq sayı" (number of rooms) is nearly fully populated for apartments (Köhnə/Yeni tikili) and Office, and 100% missing for "Obyekt"/"Qaraj"/"Torpaq", but for "Həyət evi"/"Bağ evi" only 2.3% is missing. This small remainder looks more like genuine MCAR (the person listing the property simply left the field blank, with no systematic cause).

## 5. Applying changes

To keep a clean before/after comparison, I keep the original df untouched and apply all changes on a copy, df_clean.

```python
df_clean = df.copy()

# "Binanın növü": almost entirely empty, not a useful colum
df_clean = df_clean.drop(columns = ["Binanın növü"])
```

```python
# featured / vip: convert to boolean flags (missing = False)
df_clean["featured_flag"] = df_clean["featured"].notnull()
df_clean["vip_flag"] = df_clean["vip"].notnull()
df_clean = df_clean.drop(columns = ["featured", "vip"])
df_clean[["featured_flag", "vip_flag"]].mean().round(3)
```

```python
# mortgage / İpoteka: duplicate pair -> single flag
df_clean["mortgage_flag"] = df_clean["mortgage"].notnull()
df_clean = df_clean.drop(columns = ["mortgage", "İpoteka"])
print(df_clean["mortgage_flag"].mean().round(3))
```

```python
# reviewing the category values before creating a binary feature
print("bill_of_sale values:")
df_clean["bill_of_sale"].value_counts(dropna = False)
```

```python
print("\nÇıxarış values:")
df_clean["Çıxarış"].value_counts(dropna = False)
```

The bill_of_sale column contains no missing values. Instead of using .notnull(), the actual category values were examined and converted into a binary flag.

```python
# Encode the actual categories into a binary flag
df_clean["bill_of_sale_flag"] = (
    df_clean["Çıxarış"]
    .str.strip()
    .str.lower()
    .map({
        "var": 1,
        "yoxdur": 0
    })
)

df_clean = df_clean.drop(columns = ["bill_of_sale", "Çıxarış"])

df_clean["bill_of_sale_flag"].value_counts()
```

```python
print(df_clean["bill_of_sale_flag"].isnull().sum())
```

```python
# repair / Təmir: keep Təmir as the primary field, fill its remaining gaps
df_clean["Təmir"] = df_clean["Təmir"].fillna("Unknown")
df_clean = df_clean.drop(columns = ["repair"])
df_clean["Təmir"].value_counts()
```

```python
# "Torpaq sahəsi": structural missingness, left as-is
print("Nulls: ", df_clean["Torpaq sahəsi"].isnull().sum())
```

```python
# hour_y: low analytical value, dropped
df_clean = df_clean.drop(columns = ["hour_y"])
print("'hour_y' dropped.")
```

```python
# shop_name/owner_name group: derive a seller_type column
def seller_type(row):
    if pd.notnull(row["shop_name"]):
        return "Agency"
    elif pd.notnull(row["owner_name"]):
        return "Individual"
    else:
        return "Unknown"
```

```python
df_clean["seller_type"] = df_clean.apply(seller_type, axis = 1)

for col in ["shop_name", "shop_title", "owner_name", "owner_title"]:
    df_clean[col] = df_clean[col].fillna("N/A")

df_clean["seller_type"].value_counts()
```

```python
# products_label: 'Regular listing' when no special tag applies
df_clean["products_label"] = df_clean["products_label"].fillna("Regular listing")
df_clean["products_label"].value_counts()
```

```python
# "Mərtəbə": structural missingness, left as-is
print("Rows intentionally left null for 'Mərtəbə':", df_clean["Mərtəbə"].isnull().sum())
```

```python
# unit_price: recalculated from the "Sahə" column (0% missing)
# "Sahə" comes in two units, 'm2' and 'sot' (for land) -> convert everything to m2 (1 sot = 100 m2)
def parse_area(val):
    if pd.isnull(val):
        return np.nan
    val = val.strip()
    if "sot" in val:
        return float(val.replace("sot", "").strip().replace(",", ".")) * 100
    return float(val.replace("m²", "").strip().replace(" ", ""))
```

```python
df_clean["Sahə_numeric"] = df_clean["Sahə"].apply(parse_area)
df_clean["unit_price_calculated"] = (df_clean["price"] / df_clean["Sahə_numeric"]).round(2)
df_clean = df_clean.drop(columns = ["unit_price"])

print("Missing rows:", df_clean["unit_price_calculated"].isnull().sum())
df_clean["unit_price_calculated"].describe()
```

```python
# Otaq sayı: median-fill only within applicable categories
applicable_categories = ["Köhnə tikili", "Yeni tikili", "Ofis", "Həyət evi/Bağ evi"]
mask = df_clean["Kateqoriya"].isin(applicable_categories)
medians = df_clean.loc[mask].groupby("Kateqoriya")["Otaq sayı"].median()
print("Median room count by category:")
medians
```

```python
for cat, med in medians.items():
    idx = df_clean[(df_clean["Kateqoriya"] == cat) & (df_clean["Otaq sayı"].isnull())].index
    df_clean.loc[idx, "Otaq sayı"] = med
```

```python
df_clean["Kateqoriya"].unique()
```

```python
df_clean[df_clean["Otaq sayı"].isnull()]["Kateqoriya"].value_counts()
```

```python
print("\nRemaining missing rows for 'Otaq sayı' (structural missing):",
      df_clean["Otaq sayı"].isnull().sum())

print(df_clean[df_clean["Otaq sayı"].isnull()]["Kateqoriya"].value_counts())
```

```python
# description: text field, fill with empty string
df_clean["description"] = df_clean["description"].fillna("")
print("Remaining missing rows for 'description':", df_clean["description"].isnull().sum())
```

## 6. Final check

After all steps, I re-check the missingness status. Aside from the structural missing values in "Torpaq sahəsi", "Mərtəbə", and the remaining "Otaq sayı" rows, almost every gap has now either been filled, converted into a flag, or removed.

```python
final_nulls = df_clean.isnull().sum()
final_nulls[final_nulls > 0]
```
